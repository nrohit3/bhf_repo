```python
dbutils.widgets.text("source_system","")
source_system=dbutils.widgets.get("source_system")
```

```python
print("source system is ",source_system)
```

```python
if not source_system:
    raise ValueError("source system widget cant be empty")
```

```python
tables_df=spark.table("ts_bhf_dev.governance.tables")
```

```python
rows=tables_df.filter(f"active_flag = true AND lower(source_system) = lower('{source_system}')").orderBy(("load_order")).collect()
```

```python
print(rows)
```

```python
tables_list=[]
for row in rows:
    table_id=row['table_id']
    table_name=row['table_name']
    source_system=row['source_system']
    source_schema=row['source_schema']
    source_table=row['source_table']
    source_path=row['source_path']
    target_layer=row['target_layer']
    raw_schema=row['raw_schema']
    cleansed_schema=row['cleansed_schema']
    curated_schema=row['curated_schema']
    active_flag=row['active_flag']
    load_order=row['load_order']
    created_at=row['created_at']
    tables_list.append({"table_id":table_id,"table_name":table_name,"source_system":source_system,"source_schema":source_schema,"source_table":source_table,"source_path":source_path,"target_layer":target_layer,"raw_schema":raw_schema,"cleansed_schema":cleansed_schema,"curated_schema":curated_schema,"active_flag":active_flag,"load_order":load_order,"created_at":created_at})
    print(table_id,table_name,source_system,source_schema,source_table,source_path,target_layer,raw_schema,cleansed_schema,curated_schema,active_flag,load_order,created_at)
```

```python
tables_list
```

```python
dbutils.jobs.taskValues.set("tables_metadata",tables_list)
```
